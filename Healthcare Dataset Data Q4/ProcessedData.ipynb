```python
import pandas as pd 
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("Healthcare_dataset.csv")
```

```python
data.info()
```

```python
#check null values 
data.isnull().sum()
```

```python
#check duplicate values 
data.duplicated().sum()
```

```python
#check for negative bill amount
negative_bill_amount=(data['Billing Amount']<0).sum()
print(negative_bill_amount)
```

```python
#WORK on copy of data 
df_clean=data.copy()
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
```

```python
#Convert admission and discharge dates to datetime format
df_clean['Date of Admission'] = pd.to_datetime(df_clean['Date of Admission'])
df_clean['Discharge Date'] = pd.to_datetime(df_clean['Discharge Date'])
#Correct negative billing values
df_clean['Billing Amount'] = df_clean['Billing Amount'].abs()
df_clean['Billing Amount'] = df_clean['Billing Amount'].round(2)
print(df_clean.head())
```

```python
#Feature engineering 
#Calculate Length of Stay (LOS) in days
df_clean['Length_of_Stay'] = (
    df_clean['Discharge Date'] - df_clean['Date of Admission']
).dt.days

age_bins = [0, 18, 35, 50, 65, 100]
age_labels = [
    'Pediatric (<18)',
    'Young Adult (18-35)',
    'Adult (36-50)',
    'Middle Aged (51-65)',
    'Senior (65+)',
]
df_clean['Age_Group'] = pd.cut(
    df_clean['Age'], bins=age_bins, labels=age_labels, right=True
)

df_clean['Daily_Billing_Rate'] = (
    df_clean['Billing Amount'] / df_clean['Length_of_Stay']
).round(2)

df_clean['Is_Abnormal_Test'] = df_clean['Test Results'].apply(
    lambda x: 1 if x == 'Abnormal' else 0
)

print(df_clean.head())
```

```python
# Aggregate financial and operational KPIs by Medical Condition
condition_summary = df_clean.groupby('Medical Condition').agg(
    Total_Admissions=('Name', 'count'),
    Avg_Billing_Amount=('Billing Amount', 'mean'),
    Total_Revenue=('Billing Amount', 'sum'),
    Avg_Length_of_Stay=('Length_of_Stay', 'mean'),
    Abnormal_Test_Rate=('Is_Abnormal_Test', 'mean')
).reset_index()

# Convert abnormal test rate to percentage
condition_summary['Abnormal_Test_Rate'] = (condition_summary['Abnormal_Test_Rate'] * 100).round(2)
condition_summary['Avg_Billing_Amount'] = condition_summary['Avg_Billing_Amount'].round(2)
condition_summary['Avg_Length_of_Stay'] = condition_summary['Avg_Length_of_Stay'].round(2)
```

```python
# Export the final clean dataset to CSV
df_clean.to_csv('Processed_Healthcare_Data.csv', index=False)

print("Processed dataset saved as'Processed_Healthcare_Data.csv'.")
```

```python
#EDA
test_counts = df_clean['Test Results'].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(
    test_counts,
    labels=test_counts.index,
    autopct='%1.1f%%',
    colors=['red','yellow','green'],
    startangle=90,
)
plt.title('Diagnostic Test Results Distribution')
plt.show()
```

```python
#analyze categorical features
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

admission_counts = df_clean['Admission Type'].value_counts()
ax1 = sns.barplot(
    x=admission_counts.index,
    y=admission_counts.values,
    ax=axes[0],
    
)
ax1.bar_label(ax1.containers[0])
axes[0].set_title('Admission Type Distribution')

condition_counts = df_clean['Medical Condition'].value_counts()
ax2 = sns.barplot(
    x=condition_counts.index,
    y=condition_counts.values,
    ax=axes[1],
    
)
ax2.bar_label(ax2.containers[0])
axes[1].set_title('Medical Condition Volume')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
